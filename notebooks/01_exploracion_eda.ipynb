```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
```

```python
ruta = "../data/raw/yellow_tripdata_2016-01.csv"

df = pd.read_csv(
    ruta,
    nrows=200_000,
)
df.head()
```

```python
df.shape
df.columns
```

```python
df[["trip_distance", "fare_amount", "passenger_count"]].describe()
```

```python
plt.figure()
df["fare_amount"].hist(bins=50)
plt.title("Distribución de tarifas")
plt.xlabel("Tarifa")
plt.ylabel("Frecuencia")
plt.show()

plt.figure()
df["trip_distance"].hist(bins=50)
plt.title("Distribución de distancias")
plt.xlabel("Distancia (millas)")
plt.ylabel("Frecuencia")
plt.show()
```
